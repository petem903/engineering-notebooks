```python
from sympy import symbols, laplace_transform, Heaviside, exp, cos

t, s, omega1, omega2 = symbols('t s omega1 omega2')

# Part I.a - Defining x(t) as shown in Figure 2

x1 = t * Heaviside(t - 1) - t * Heaviside(t - 2)
x2 = Heaviside(t - 2) - Heaviside(t - 4)

X1_s = laplace_transform(x1, t, s, noconds=True)
X2_s = laplace_transform(x2, t, s, noconds=True)
X_s = X1_s + X2_s

print("X1(s):", X1_s)
print("X2(s):", X2_s)
print("X(s) = X1(s) + X2(s):", X_s)

# Part I.b - Calculating X(s) for given x(t) expressions

x_expr1 = (2 * exp(-t) - exp(-t)) * Heaviside(t)
X_expr1_s = laplace_transform(x_expr1, t, s, noconds=True)
print("\nX(s) for x(t) = (2*e^(-t) - e^(-t))u(t):", X_expr1_s)

x_expr2 = Heaviside(t) - Heaviside(t - 1)
X_expr2_s = laplace_transform(x_expr2, t, s, noconds=True)
print("X(s) for x(t) = u(t) - u(t - 1):", X_expr2_s)

x_expr3 = t * exp(-t) * Heaviside(t)
X_expr3_s = laplace_transform(x_expr3, t, s, noconds=True)
print("X(s) for x(t) = t * e^(-t) * u(t):", X_expr3_s)

x_expr4 = cos(omega1 * t) * cos(omega2 * t) * Heaviside(t)
X_expr4_s = laplace_transform(x_expr4, t, s, noconds=True)
print("X(s) for x(t) = cos(omega1*t) * cos(omega2*t) * u(t):", X_expr4_s)
```

```python
from sympy import symbols, integrate, pi, cos, sin, exp, summation

t, n = symbols('t n')

T0 = pi
omega0 = 2 * pi / T0
x_t = exp(-t / 2)

# Part III - Fourier Series Coefficients

# a0
a0 = (1 / T0) * integrate(x_t, (t, 0, T0))
print("a0:", a0.evalf())

# an
an = (2 / T0) * integrate(x_t * cos(n * omega0 * t), (t, 0, T0))
print("an:", an.simplify())

# bn
bn = (2 / T0) * integrate(x_t * sin(n * omega0 * t), (t, 0, T0))
print("bn:", bn.simplify())

# Fourier Series x(t)
xt = a0 + summation(an * cos(n * omega0 * t) + bn * sin(n * omega0 * t), (n, 1, 20))

# Plot x(t) approximation
from sympy.plotting import plot
plot(xt, (t, -0.1, 4))
```

```python
from sympy import symbols, integrate, pi, cos, sin, Piecewise, summation
from sympy.plotting import plot

t, n = symbols('t n')
T0 = 2 * pi  

x_t = Piecewise((1, (t >= 0) & (t < pi)), (0, (t >= pi) & (t < 2 * pi)))

# Problems III - Fourier Series Coefficients

# a0
a0 = (1 / T0) * integrate(x_t, (t, 0, T0))
print("a0:", a0.evalf())

# an
an = (2 / T0) * integrate(x_t * cos(n * (2 * pi / T0) * t), (t, 0, T0))
print("an:", an.simplify())

# bn
bn = (2 / T0) * integrate(x_t * sin(n * (2 * pi / T0) * t), (t, 0, T0))
print("bn:", bn.simplify())

# Fourier Series x(t)
xt = a0 + summation(an * cos(n * (2 * pi / T0) * t) + bn * sin(n * (2 * pi / T0) * t), (n, 1, 20))

# Plot x(t) approximation
plot(xt, (t, -3 * pi, 3 * pi), ylabel="x(t)")
```

```python
from sympy import symbols, I, Abs, arg, sqrt
import numpy as np
import matplotlib.pyplot as plt

omega, a = symbols('omega a', real=True)

X_omega = 1 / (I * omega + a)

X_mag = Abs(X_omega)
X_mag_func = X_mag.subs(a, 1).evalf()

X_phase = arg(X_omega)
X_phase_func = X_phase.subs(a, 1).evalf()

X_mag_func_np = lambda w: 1 / np.sqrt(w**2 + 1)
X_phase_func_np = lambda w: -np.arctan(w / 1)

omega_vals = np.linspace(-10, 10, 400)
plt.figure()
plt.plot(omega_vals, X_mag_func_np(omega_vals))
plt.title("|X(omega)|")
plt.xlabel("omega")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()

plt.figure()
plt.plot(omega_vals, X_phase_func_np(omega_vals))
plt.title("∠X(omega)")
plt.xlabel("omega")
plt.ylabel("Phase (radians)")
plt.grid(True)
plt.show()
```

```python
from sympy import symbols, fourier_transform, Abs, arg
import numpy as np
import matplotlib.pyplot as plt

# Define symbols
t, omega = symbols('t omega')

# Define rect function for x(t)
x_t = 1 * (Abs(t) <= 1/2)

# Fourier Transform of x(t)
X_omega = fourier_transform(x_t, t, omega)
print("X(omega):", X_omega)

# Magnitude and Phase
X_mag = Abs(X_omega)
X_phase = arg(X_omega)

# Convert to functions for plotting
X_mag_func = lambda w: np.abs(np.sinc(w / np.pi))
X_phase_func = lambda w: np.zeros_like(w)  # Phase is zero for rect function

# Plot Magnitude
omega_vals = np.linspace(-10, 10, 400)
plt.figure()
plt.plot(omega_vals, X_mag_func(omega_vals))
plt.title("|X(omega)|")
plt.xlabel("omega")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()

# Plot Phase
plt.figure()
plt.plot(omega_vals, X_phase_func(omega_vals))
plt.title("∠X(omega)")
plt.xlabel("omega")
plt.ylabel("Phase (radians)")
plt.grid(True)
plt.show()
```
