### Problem 5) 5 Points

* Find the root of the equation $x^2 - 4x - 7 = 0$ near x = 5 using Newton-Raphson Method 

$$x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}$$
$$.$$
$$.$$
$$.$$
$$x_{n+1} = x_{n} - \frac{f(x_n)}{f'(x_n)}$$ 


````python
def f(x):
    return x**2 - 4*x - 7

def f_prime(x):
    return 2*x - 4

def newton_raphson(x0, tol=1e-6):
    x = x0
    while abs(f(x)) > tol:
        x = x - f(x)/f_prime(x)
    return x

x0 = 5
root = newton_raphson(x0)
print(f"Root: {root}")
````

### Problem 6) 10 Points

    Exam I -> 15%
    Exam II -> 20%
    Exam III ->25%
    HW (5) -> 20%

    Project -> 10

Use the grade calculator in Problem 4 to display letter grade. 


````python
def grade_calculator(exam1, exam2, exam3, hw, quiz, project):
    total = (exam1 * 0.15) + (exam2 * 0.20) + (exam3 * 0.25) + (hw * 0.20) + (quiz * 0.10) + (project * 0.10)
    return total

def letter_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

exam1 = 85
exam2 = 90
exam3 = 88
hw = 80
quiz = 75
project = 92

final_score = grade_calculator(exam1, exam2, exam3, hw, quiz, project)
print(f"Final Score: {final_score}")
print(f"Letter Grade: {letter_grade(final_score)}")
````

### Problem 7) 10 Points

Write a Python program to guess a number between 0 to 100. 

- Create a number to guess
    - static
    - random
- Ask user enter a number
- Check if that is a number correct
 - If it is not ask another number
 - If it is correct display the number and number of guess

````python
import random

def guess_number():
    number_to_guess = random.randint(0, 100)
    guesses = 0
    while True:
        user_guess = int(input("Enter your guess (0-100): "))
        guesses += 1
        if user_guess == number_to_guess:
            print(f"Correct! The number was {number_to_guess}. You guessed it in {guesses} tries.")
            break
        elif user_guess < number_to_guess:
            print("Too low!")
        else:
            print("Too high!")

guess_number()
````

### Problem 8) 10 Points

* Create a random discrete signal x[n] has a size 100. Values of x[n] will be between (-1,1)
* Find x[2n] and x[4n] (Downsample)
* Using x[2n] recover x[n]
 * First fill with zeros
 * Calculate zero x[i] using average of (x[i+1]+x[i-1])/2

````python
import random

x = [random.uniform(-1, 1) for _ in range(100)]

x_2n = x[::2]
x_4n = x[::4]

x_recovered = [0] * 100
x_recovered[::2] = x_2n

for i in range(1, len(x_recovered), 2):
    if i + 1 < len(x_recovered):
        x_recovered[i] = (x_recovered[i - 1] + x_recovered[i + 1]) / 2

# Print results
print(f"x[2n]: {x_2n}")
print(f"x[4n]: {x_4n}")
print(f"Recovered x[n]: {x_recovered}")
````

### Problem 10) 10 Points

a) Write a python function that calculates $e^x$ using the Taylor series representation shown below

$$e^x = \sum_{n=0}^{\infty}\frac{1}{n!}x^n = \sum_{n=0}^{\infty}\frac{x^n}{n!}$$

Verify your function operation using python ```math``` library $e^x$ function for 20 random x values. Your error must be less than $10^{-6}$ 

b) Write a python function that calculates $e^{-x}$ using the Taylor series representation for 10 terms (n=0,1,2,...,9) shown below

$$e^{-x} = \sum_{n=0}^{\infty}\frac{1}{n!}(-x)^n = \sum_{n=0}^{\infty}\frac{(-1)^nx^n}{n!}$$

Verify your function operation using python ```math``` library $e^x$ function for 20 random x values. Show your error. 

c) Write a python function that calculates $sin(x)$ using the Taylor series representation shown below

$$sin(x) = \sum_{n=0}^{n}\frac{(-1)^nx^{2n+1}}{(2n+1)!}$$

Verify your function operation using python ```math``` library $sin(x)$ function for 20 random x values. Show your error. 

d) Write a python function that calculates $cos(x)$ using the Taylor series representation shown below

$$cos(x) = \sum_{n=0}^{n}\frac{(-1)^nx^{2n}}{(2n)!}$$

Verify your function operation using python ```math``` library $cos(x)$ function for 20 random x values. Show your error. 

````python
import math

# a) Taylor series for e^x
def exp_taylor(x, terms=10):
    result = 0
    for n in range(terms):
        result += (x**n) / math.factorial(n)
    return result

# b) Taylor series for e^(-x)
def exp_neg_taylor(x, terms=10):
    result = 0
    for n in range(terms):
        result += ((-1)**n * x**n) / math.factorial(n)
    return result

# c) Taylor series for sin(x)
def sin_taylor(x, terms=10):
    result = 0
    for n in range(terms):
        result += ((-1)**n * x**(2*n+1)) / math.factorial(2*n+1)
    return result

# d) Taylor series for cos(x)
def cos_taylor(x, terms=10):
    result = 0
    for n in range(terms):
        result += ((-1)**n * x**(2*n)) / math.factorial(2*n)
    return result

# Verification with math library
import random

for _ in range(20):
    x = random.uniform(-10, 10)
    
    # Check exp(x)
    approx_exp = exp_taylor(x)
    actual_exp = math.exp(x)
    print(f"exp({x}): Taylor = {approx_exp}, math = {actual_exp}, error = {abs(approx_exp - actual_exp)}")
    
    # Check exp(-x)
    approx_exp_neg = exp_neg_taylor(x)
    actual_exp_neg = math.exp(-x)
    print(f"exp(-{x}): Taylor = {approx_exp_neg}, math = {actual_exp_neg}, error = {abs(approx_exp_neg - actual_exp_neg)}")
    
    # Check sin(x)
    approx_sin = sin_taylor(x)
    actual_sin = math.sin(x)
    print(f"sin({x}): Taylor = {approx_sin}, math = {actual_sin}, error = {abs(approx_sin - actual_sin)}")
    
    # Check cos(x)
    approx_cos = cos_taylor(x)
    actual_cos = math.cos(x)
    print(f"cos({x}): Taylor = {approx_cos}, math = {actual_cos}, error = {abs(approx_cos - actual_cos)}")
````
