```python
# Problem 1: Resistor Color Code Calculations

# Part (a): Function to Calculate Resistor Value from Color Code


color_code_digits = {
    "Black": 0, "Brown": 1, "Red": 2, "Orange": 3, "Yellow": 4,
    "Green": 5, "Blue": 6, "Violet": 7, "Gray": 8, "White": 9
}

color_code_multipliers = {
    "Black": 1, "Brown": 10, "Red": 100, "Orange": 1_000, "Yellow": 10_000,
    "Green": 100_000, "Blue": 1_000_000, "Gold": 0.1, "Silver": 0.01
}

color_code_tolerances = {
    "Brown": 1, "Red": 2, "Green": 0.5, "Blue": 0.25, "Violet": 0.1,
    "Gray": 0.05, "Gold": 5, "Silver": 10
}


def Color_Res(band1, band2, multiplier_band, tolerance_band):
    digit1 = color_code_digits.get(band1, -1)
    digit2 = color_code_digits.get(band2, -1)
    multiplier = color_code_multipliers.get(multiplier_band, -1)
    tolerance = color_code_tolerances.get(tolerance_band, -1)

   
    if -1 in (digit1, digit2, multiplier, tolerance):
        return "Invalid color code"

   
    resistor_value = (digit1 * 10 + digit2) * multiplier
    return f"{resistor_value} ± {tolerance}% Ohms"


print(Color_Res("Brown", "Black", "Red", "Gold"))  # Expected Output: 1000 ± 5% Ohms


# Part (b): Function to Calculate Color Code for Given Resistor Value

def Res_Col(resistor_value, tolerance):
    digit_colors = {v: k for k, v in color_code_digits.items()}
    multiplier_colors = {v: k for k, v in color_code_multipliers.items()}
    tolerance_colors = {v: k for k, v in color_code_tolerances.items()}

    resistor_str = str(int(resistor_value))
    first_digit = int(resistor_str[0])
    second_digit = int(resistor_str[1])
    multiplier = 10 ** (len(resistor_str) - 2)

    band1 = digit_colors.get(first_digit, "Invalid")
    band2 = digit_colors.get(second_digit, "Invalid")
    multiplier_band = multiplier_colors.get(multiplier, "Invalid")
    tolerance_band = tolerance_colors.get(tolerance, "Invalid")

    if "Invalid" in (band1, band2, multiplier_band, tolerance_band):
        return "Invalid resistor value"

    return f"{band1},{band2},{multiplier_band},{tolerance_band}"

print(Res_Col(1000, 5))  # Expected Output: Brown,Black,Red,Gold


# Part (c): Test Script to Verify Part (a) and Part (b)

import random


random_resistors = [(random.choice([10, 100, 1000, 10000, 100000]), random.choice([1, 2, 5, 10])) for _ in range(10)]
color_codes = []
calculated_values = []


for resistor_value, tolerance in random_resistors:
    color_code = Res_Col(resistor_value, tolerance)
    color_codes.append(color_code)
    

    bands = color_code.split(",")
    calculated_value = Color_Res(bands[0], bands[1], bands[2], bands[3])
    calculated_values.append(calculated_value)


print(f"{'Resistor Value (Ohms)':<20}{'Tolerance (%)':<15}{'Color Code':<25}{'Calculated Value'}")
print("-" * 75)
for i in range(10):
    print(f"{random_resistors[i][0]:<20}{random_resistors[i][1]:<15}{color_codes[i]:<25}{calculated_values[i]}")
```

```python
# Problem 2: Matrix Operations

# Part (a): Matrix Multiplication without Numpy

def matrix_multiply(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            C[i][j] = sum(A[i][k] * B[k][j] for k in range(n))
    
    return C

A = [[i + j for j in range(4)] for i in range(4)]
B = [[i * j for j in range(4)] for i in range(4)]
print("Matrix A * B without Numpy (n=4):")
print(matrix_multiply(A, B))


# Part (b): Matrix Multiplication using Numpy

import numpy as np

def matrix_multiply_numpy(A, B):
    A_np = np.array(A)
    B_np = np.array(B)
    return np.dot(A_np, B_np)

print("Matrix A * B with Numpy (n=4):")
print(matrix_multiply_numpy(A, B))


# Part (c): Element-by-Element Comparison

def compare_matrices(C1, C2):
    n = len(C1)
    for i in range(n):
        for j in range(n):
            if C1[i][j] != C2[i][j]:
                return False
    return True

C1 = matrix_multiply(A, B)
C2 = matrix_multiply_numpy(A, B)
print("Matrices are equal:", compare_matrices(C1, C2))


# Part (d): Performance Comparison with Timeit

import timeit
import matplotlib.pyplot as plt

matrix_sizes = [4, 6, 10, 20]
times_without_numpy = []
times_with_numpy = []

for n in matrix_sizes:
    A = [[i + j for j in range(n)] for i in range(n)]
    B = [[i * j for j in range(n)] for i in range(n)]
    
    time_without_numpy = timeit.timeit(lambda: matrix_multiply(A, B), number=10)
    times_without_numpy.append(time_without_numpy)
    
    time_with_numpy = timeit.timeit(lambda: matrix_multiply_numpy(A, B), number=10)
    times_with_numpy.append(time_with_numpy)

plt.plot(matrix_sizes, times_without_numpy, label='Without Numpy')
plt.plot(matrix_sizes, times_with_numpy, label='With Numpy')
plt.xlabel("Matrix Size (n x n)")
plt.ylabel("Time (seconds)")
plt.title("Matrix Multiplication Performance Comparison")
plt.legend()
plt.show()
```

```python
# Problem 3: Signals and Systems

import numpy as np
import matplotlib.pyplot as plt

# Part (a): Continuous-Time Signal
frequency = 1000 
omega = 2 * np.pi * frequency  
time = np.linspace(0, 3 / frequency, 500)  

x_t = 3 * np.cos(omega * time)


plt.figure(figsize=(10, 4))
plt.plot(time, x_t, label='x(t) = 3cos(ωt)')
plt.title("Continuous-Time Signal x(t) for 3 Cycles")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()

# Part (b): Discrete-Time Signal x[n]

N = 16 
fs = 8000  
T = 1 / fs  

n = np.arange(N)
t_samples = n * T  


x_n = 3 * np.cos(omega * t_samples)


print("Discrete samples x[n]:", x_n)

# Part (c): Plot Continuous and Discrete Signals Together
plt.figure(figsize=(10, 4))
plt.plot(time, x_t, label='x(t) = 3cos(ωt)')
plt.stem(t_samples, x_n, linefmt='r-', markerfmt='ro', basefmt=" ", label='x[n]', use_line_collection=True)
plt.title("Continuous-Time Signal x(t) and Discrete Signal x[n]")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()

# Part (d): Discrete Fourier Transform (DFT) of x[n]
X_m = np.fft.fft(x_n)


magnitude = np.abs(X_m)
phase = np.angle(X_m)

plt.figure(figsize=(10, 6))


plt.subplot(2, 1, 1)
plt.stem(n, magnitude, basefmt=" ", use_line_collection=True)
plt.title("Magnitude of DFT of x[n]")
plt.xlabel("Frequency Index (m)")
plt.ylabel("|X[m]|")


plt.subplot(2, 1, 2)
plt.stem(n, phase, basefmt=" ", use_line_collection=True)
plt.title("Phase of DFT of x[n]")
plt.xlabel("Frequency Index (m)")
plt.ylabel("∠X[m] (radians)")

plt.tight_layout()
plt.show()

# Part (e): Inverse DFT to Reconstruct x[n]
def inverse_dft(X_m):
    N = len(X_m)
    x_n_reconstructed = []
    for n in range(N):
        x_n_value = sum(X_m[m] * np.exp(1j * 2 * np.pi * m * n / N) for m in range(N)) / N
        x_n_reconstructed.append(x_n_value.real)  # Take the real part only
    return np.array(x_n_reconstructed)

x_n_reconstructed = inverse_dft(X_m)

print("Reconstructed x[n]:", x_n_reconstructed)

# Part (f): Plot Original and Reconstructed x[n] for Comparison
plt.figure(figsize=(10, 4))
plt.stem(n, x_n, linefmt='b-', markerfmt='bo', basefmt=" ", label='Original x[n]', use_line_collection=True)
plt.stem(n, x_n_reconstructed, linefmt='r--', markerfmt='ro', basefmt=" ", label='Reconstructed x[n]', use_line_collection=True)
plt.title("Comparison of Original x[n] and Reconstructed x[n] Using IDFT")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Problem 5: Solving Linear Systems and Complex Number Operations

# Part (a): Solving Linear Equations with Gaussian Elimination
import numpy as np

def LinSolve(A, b, n):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    
    for i in range(n):
        factor = A[i][i]
        for j in range(i, n):
            A[i][j] /= factor
        b[i] /= factor
        
        for k in range(i + 1, n):
            factor = A[k][i]
            for j in range(i, n):
                A[k][j] -= factor * A[i][j]
            b[k] -= factor * b[i]
    
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = b[i]
        for j in range(i + 1, n):
            x[i] -= A[i][j] * x[j]
    
    result = ', '.join([f"x{i+1}={x[i]:.2f}" for i in range(n)])
    return result


A = [
    [2, 1, 1],
    [1, 3, -1],
    [1, 1, 1]
]
b = [3, 7, 1]

print("Part (a): Solution for 3-variable system")
print(LinSolve(A, b, 3))


# Part (b): Operations with Complex Numbers

import cmath
import math

w1 = 3 + 4j
r_w2 = 2
theta_w2 = math.pi / 4
w2 = r_w2 * (math.cos(theta_w2) + 1j * math.sin(theta_w2))  

magnitude_w1 = abs(w1)
phase_w1 = cmath.phase(w1)
print("\nPart (i): w1 in polar form")
print(f"w1 = {magnitude_w1:.2f} * e^(j{phase_w1:.2f})")


print("\nPart (ii): w2 in rectangular form")
print(f"w2 = {w2.real:.2f} + j{w2.imag:.2f}")

magnitude_w1_squared = magnitude_w1 ** 2
magnitude_w2_squared = abs(w2) ** 2
print("\nPart (iii): |w1|^2 and |w2|^2")
print(f"|w1|^2 = {magnitude_w1_squared:.2f}")
print(f"|w2|^2 = {magnitude_w2_squared:.2f}")

w1_plus_w2 = w1 + w2
print("\nPart (iv): w1 + w2 in rectangular form")
print(f"w1 + w2 = {w1_plus_w2.real:.2f} + j{w1_plus_w2.imag:.2f}")

w1_minus_w2 = w1 - w2
magnitude_w1_minus_w2 = abs(w1_minus_w2)
phase_w1_minus_w2 = cmath.phase(w1_minus_w2)
print("\nPart (v): w1 - w2 in polar form")
print(f"w1 - w2 = {magnitude_w1_minus_w2:.2f} * e^(j{phase_w1_minus_w2:.2f})")

w1_times_w2 = w1 * w2
print("\nPart (vi): w1 * w2 in rectangular form")
print(f"w1 * w2 = {w1_times_w2.real:.2f} + j{w1_times_w2.imag:.2f}")

w1_divided_w2 = w1 / w2
magnitude_w1_divided_w2 = abs(w1_divided_w2)
phase_w1_divided_w2 = cmath.phase(w1_divided_w2)
print("\nPart (vii): w1 / w2 in polar form")
print(f"w1 / w2 = {magnitude_w1_divided_w2:.2f} * e^(j{phase_w1_divided_w2:.2f})")
```

```python
# Problem 6: Newton-Raphson Method for Finding Roots

import math

# Part (a): Newton-Raphson Method Based on Iterations

def newton_raphson_iter(f, df, x0, iterations):
    """
    Applies the Newton-Raphson method for a specified number of iterations.
    """
    results = []
    x = x0
    for i in range(iterations):
        fx = f(x)
        results.append((i, x, fx)) 
        x = x - fx / df(x) 
    return results

def f(x):
    return math.cos(x) - x

def df(x):
    return -math.sin(x) - 1

x0 = 1
iterations = 5

results = newton_raphson_iter(f, df, x0, iterations)

print("Iteration\tX\t\tf(X)")
for i, x, fx in results:
    print(f"{i}\t\t{x:.6f}\t{fx:.6e}")


# Part (b): Newton-Raphson Method with Tolerance for Convergence

def newton_raphson_tol(f, df, x0, tolerance):
    """
    Applies Newton-Raphson method until a specified tolerance level is met.
    """
    results = []
    x = x0
    iteration = 0
    while True:
        fx = f(x)
        results.append((iteration, x, fx))
        
        x_new = x - fx / df(x)
        if abs(x_new - x) < tolerance:
            break
        x = x_new
        iteration += 1

    return results

tolerance = 1e-6
results_tol = newton_raphson_tol(f, df, x0, tolerance)

print("\nIteration\tX\t\tf(X)")
for i, x, fx in results_tol:
    print(f"{i}\t\t{x:.6f}\t{fx:.6e}")


# Part (c): Additional Test Functions for Newton-Raphson
def f2(x):
    return x * math.tan(x) - 1

def df2(x):
    return math.tan(x) + x / (math.cos(x)**2)

def f3(x):
    return x**3 - 10 * x**2 + 5

def df3(x):
    return 3 * x**2 - 20 * x


x0_test2 = 0.5  
x0_test3 = 0.1 

print("\nTesting f(x) = x * tan(x) - 1 with Newton-Raphson method")
results_test2 = newton_raphson_iter(f2, df2, x0_test2, iterations=5)
for i, x, fx in results_test2:
    print(f"{i}\t\t{x:.6f}\t{fx:.6e}")

print("\nTesting f(x) = x^3 - 10x^2 + 5 with Newton-Raphson method")
results_test3 = newton_raphson_iter(f3, df3, x0_test3, iterations=5)
for i, x, fx in results_test3:
    print(f"{i}\t\t{x:.6f}\t{fx:.6e}")


# Part (d): Write Newton-Raphson Results to a File

def write_results_to_file(filename, results):
    """
    Writes Newton-Raphson results to a file in a table format.
    """
    with open(filename, 'w') as file:
        file.write("Iteration\tX\t\tf(X)\n")
        for i, x, fx in results:
            file.write(f"{i}\t\t{x:.6f}\t{fx:.6e}\n")

write_results_to_file("newton_raphson_results.txt", results)
print("\nResults written to newton_raphson_results.txt")
```
